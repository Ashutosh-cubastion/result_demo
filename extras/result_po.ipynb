```python
def get_merit(candidate_df):    
    
    candidate_df['merit'] = None
    
    candidates =  candidate_df[
    (candidate_df['total'].notnull()) &
    (candidate_df['rejection_provision'].astype(str).isin(('Q','W','TU','WWV')))
    ].sort_values(by = ['total', 'paper2', 'dob', 'name'], ascending=[False, False, True, True])
    
    candidates['merit'] = range(1, len(candidates) + 1)

    rollno_to_merit = candidates.set_index('rollno')['merit'].to_dict()
    candidate_df['merit'] = candidate_df['rollno'].map(rollno_to_merit)
    
    return "{flag_name} updated successfully for {merit} candidates".format(
        flag_name='merit',
        merit=len(candidates))
```

#### CODE

```python
def get_cutoff(cutoff_df, candidate_df, post):

    if post == 'CAPF':
        cutoff = 'cutoff_flag_cpo'
        cat1 = 'cat1_cpo'
        df_cutoff = cutoff_df[cutoff_df['sub_category'].isnull()]
        sorted_candidates = candidate_df[candidate_df['merit'].notnull()].sort_values(by='merit')

    elif post == 'DP':
        cutoff = 'cutoff_flag_dp'
        cat1 = 'cat1_dp'
        df_cutoff = cutoff_df[cutoff_df['sub_category'].isnull()]
        sorted_candidates = candidate_df[candidate_df['merit'].notnull()].sort_values(by='merit')

    elif post == 'ESM_SPL':
        cutoff = 'cutoff_flag_spclexs_dp'
        cat1 = 'cat1_dp'
        df_cutoff = cutoff_df[cutoff_df['sub_category'].notnull()]
        sorted_candidates = candidate_df[(candidate_df['merit'].notnull()) & (candidate_df['isexs_specialquota']== 'Yes')].sort_values(by='merit')

    candidate_df[cutoff] = None
    cutoff_dict_m = {}
    cutoff_dict_f = {}

    for index, row in df_cutoff.iterrows():
        category = int(row['category'])
        gender = int(row['gender'])
        paper1 = float(row['paper1'])
        paper2 = float(row['total'])
        if gender  == 1:
            cutoff_dict_f[category] = [paper1, paper2, gender]
        else:
            cutoff_dict_m[category] = [paper1, paper2, gender]
    
    
    def calculate_cutoff(candidate):
        cutoff = ''

        if candidate['gender'] == 1:
            cutoff_dict = cutoff_dict_f
        
        else:
            cutoff_dict = cutoff_dict_m
    
        if (candidate['paper1'] >= cutoff_dict[9][0]) & (candidate['p2_score'] >= cutoff_dict[9][1]) & (candidate['gender'] == cutoff_dict[9][2]):
            cutoff += '9'

        for cat in [0, 1, 2, 6]:
            if (candidate[cat1] == cat) & (cat in cutoff_dict):
                if (candidate['paper1'] >= cutoff_dict[cat][0]) & (candidate['p2_score'] >= cutoff_dict[cat][1]) & (candidate['gender'] == cutoff_dict[cat][2]):
                    cutoff += str(cat)

        if (candidate['cat2'] == 3) & (3 in cutoff_dict):
            if (candidate['paper1'] >= cutoff_dict[3][0]) & (candidate['p2_score'] >= cutoff_dict[3][1]) & (candidate['gender'] == cutoff_dict[3][2]):
                cutoff += '3'

        for cat in [4, 5, 7, 8]:
            if (candidate['cat3'] == cat) & (cat in cutoff_dict):
                if (candidate['paper1'] >= cutoff_dict[cat][0]) & (candidate['p2_score'] >= cutoff_dict[cat][1]) & (candidate['gender'] == cutoff_dict[cat][2]):
                    cutoff += str(cat)

        return cutoff

    sorted_candidates[cutoff] = sorted_candidates.apply(lambda row: calculate_cutoff(row), axis=1)
    candidate_df.update(sorted_candidates)

    return f"{cutoff} updated successfully"
```

```python
def convert_service_period(df, column_name):
    def extract_values(text):
        if pd.isna(text):
            return None, None, None

        years, months, days = 0, 0, 0
        matches = re.findall(r'(\d+)\s*(Year[s]*)', text)
        if matches:
            years = int(matches[0][0])

        matches = re.findall(r'(\d+)\s*(Month[s]*)', text)
        if matches:
            months = int(matches[0][0])

        matches = re.findall(r'(\d+)\s*(Day[s]*)', text)
        if matches:
            days = int(matches[0][0])

        return years, months, days

    extracted_data = df[column_name].apply(extract_values)
    df[['exsm_yrs', 'exsm_months', 'exsm_days']] = pd.DataFrame(extracted_data.tolist(), index=df.index)

convert_service_period(candidate_df, 'service_period')
```

```python
def get_dob_flag(candidate_df, arc_df, medium):
    
    if medium == 'CAPF':
        cat1 = 'cat1_cpo'
        dob_flag = 'dob_flag_cpo'
    
    elif medium == 'DP':
        cat1 = 'cat1_dp'
        dob_flag = 'dob_flag_dp'
    
    dob_to_date = pd.to_datetime('08/02/2004', format="%m/%d/%Y")
    dob_from_date = pd.to_datetime('08/01/1999', format="%m/%d/%Y")
    
    arc_year_dict = dict(zip(arc_df['arc_code'].astype(str), arc_df['arc_year']))

    candidate_df[dob_flag] = ''

    def calculate_verification_flag(row):
        dob = row['dob']
        verification_flag = ''
        
        if dob >= dob_to_date:
            verification_flag = 'U'
        
        elif dob_from_date < dob < dob_to_date:
            verification_flag = '9'
        
        elif dob <= dob_from_date:
            
            if row['cat2']=='' and (row[cat1]=='9' or row[cat1]=='0'):
                verification_flag = '99'
                return verification_flag
            
            agerelax_code = row['agerelax_code']
            
            if pd.isna(agerelax_code) or agerelax_code == '':
                agerelax_code = '99'
            
            if isinstance(agerelax_code, str) and len(agerelax_code.strip()) == 1:
                agerelax_code = agerelax_code.strip()
            
            age_rlx_year = arc_year_dict.get(agerelax_code, 0)
            
            if agerelax_code == '6':
                ex_service_years = row['exsm_yrs']
                ex_service_months = row['exsm_months']
                ex_service_days = row['exsm_days']
                age_rlx_year += int(ex_service_years) if ex_service_years else 0
                new_dob = dob + pd.DateOffset(years=age_rlx_year, months=ex_service_months, days=ex_service_days)
            
            else:
                extract_year = dob.year
                extract_month = dob.month
                extract_day = dob.day
                new_year = extract_year + int(age_rlx_year)
                
                if extract_month == 2 and extract_day == 29:
                    extract_day = 28
            
                new_dob = pd.to_datetime(f"{new_year}-{extract_month}-{extract_day}", format='%Y-%m-%d')
                
            if new_dob > dob_from_date:
                if len(agerelax_code) < 2:
                    verification_flag = '0' + agerelax_code
                else:
                    verification_flag = agerelax_code
            else:
                verification_flag = '99'

        return verification_flag

    sorted_candidates = candidate_df[(candidate_df['merit'].notnull())]
    
    sorted_candidates[dob_flag] = sorted_candidates.apply(calculate_verification_flag, axis=1)

    candidate_df.update(sorted_candidates)
    
    return f"{dob_flag} updated successfully"
```

```python
def get_catsel_dob(candidate_df, arc_df, medium):
    dob_from_date = pd.to_datetime('08/01/1999', format="%m/%d/%Y")
    
    if medium == 'CAPF':
        cat1 = 'cat1_cpo'
        dob_flag = 'dob_flag_cpo'
        catsel_dob_flag = 'catsel_dob_cpo'
    elif medium == 'DP':
        cat1 = 'cat1_dp'
        dob_flag = 'dob_flag_dp'
        catsel_dob_flag = 'catsel_dob_dp'

    candidate_df[catsel_dob_flag] = ''
    
    filtered_and_sorted_candidates = candidate_df[(candidate_df['merit'].notnull()) 
    & (candidate_df[dob_flag].notnull()) 
    & (candidate_df[dob_flag] != '')]
    
    def calculate_catsel_dob(row, dob_flag_name):
        dob_flag = row[dob_flag_name]
       
        if dob_flag is not None and dob_flag != '' and dob_flag != '9':
            matching_rows = arc_df[arc_df['arc_code'] == dob_flag]
            if not matching_rows.empty:
                catsel = matching_rows.iloc[0]
                catsel_cat1 = catsel['arc_cat1'].split(',')
                catsel_cat2 = catsel['arc_cat2'].split(',')
                catsel_gender = catsel['arc_gender'].split(',')
                condition = (str(row[cat1]) in catsel_cat1) & (str(int(row['gender'])) in catsel_gender)
                catsel_dob = ''
                if row['cat2'] == '3' and row['cat2'] in catsel_cat2:
                    catsel_dob = row['cat2']
                else:
                    if pd.isnull(catsel['arc_year_against_ur']):
                        catsel['arc_year_against_ur'] = 0
                    dob_datetime = pd.to_datetime(row['dob'])
                    extract_year = dob_datetime.year
                    extract_month = dob_datetime.month
                    extract_day = dob_datetime.day
                    if (extract_day == 29) and (extract_month == 2):
                        extract_day = 28
                    new_year = extract_year + int(catsel['arc_year_against_ur'])
                    new_dob = pd.to_datetime(f'{new_year}-{extract_month}-{extract_day}')
                    catsel_dob_decision = new_dob > dob_from_date
                    
                    catsel_dob = '9' if catsel_dob_decision else row[cat1]
                    
                return catsel_dob if condition else ''
            else:
                return ''
        else:
            return dob_flag
    
    filtered_and_sorted_candidates[catsel_dob_flag] = filtered_and_sorted_candidates.apply(calculate_catsel_dob, args = (dob_flag, ), axis=1)
    candidate_df.update(filtered_and_sorted_candidates)
    
    print('Successfully updated ' + catsel_dob_flag)
```

```python
def get_catsel(candidate_df, post):

    if post == 'CAPF':
        cutoff_flag = 'cutoff_flag_cpo'
        cat1 = 'cat1_cpo'
        catsel_flag = 'catsel_cpo'
        catsel_dob_flag = 'catsel_dob_cpo'
    
    elif post == 'DP':
        cutoff_flag = 'cutoff_flag_dp'
        cat1 = 'cat1_dp'
        catsel_flag = 'catsel_dp'
        catsel_dob_flag = 'catsel_dob_dp'
    
    elif post == 'ESM_SPL':
        cutoff_flag = 'cutoff_flag_spclexs_dp'
        cat1 = 'cat1_dp'
        catsel_flag = 'catsel_spclexs_dp'
        catsel_dob_flag = 'catsel_dob_dp'
        
    candidate_df[catsel_flag] = ''
    
    candidates = candidate_df[(candidate_df['merit'].notnull()) & (candidate_df[cutoff_flag].notnull())].sort_values(by='rollno')
    
    def update_cutoff_flag(candidate_record):
        if candidate_record['cat2'] == '3' and candidate_record['exs_reservation'] == 'No':
            candidate_record[cutoff_flag] = candidate_record[cutoff_flag].replace('3', '').strip()

        if candidate_record[catsel_dob_flag] in ['1', '2', '6', '4', '5', '7', '8']:
            return candidate_record[cutoff_flag].replace('9', '').strip()
        elif pd.isna(candidate_record[catsel_dob_flag]) or candidate_record[catsel_dob_flag] == '':
            return ''
        else:
            return candidate_record[cutoff_flag]
    
    candidate_df[catsel_flag] = candidates.apply(update_cutoff_flag, axis=1)

    return 'Successfully updated '+ catsel_flag
```

```python
def update_allocation_dp(candidates_df, category, post, gender, subcat, allocated_against_ur, candidate, vacancy_dict):
                        
    candidates_df.loc[candidate.name, 'allocated_category'] = category
    candidates_df.loc[candidate.name, 'allocated_post'] = post
    candidates_df.loc[candidate.name, 'allocated_sub_category'] = subcat
    candidates_df.loc[candidate.name, 'allocated_against_ur'] = allocated_against_ur
    
    return True
```

```python
def update_allocation_cpo(candidates_df, category, post, gender, subcat, allocated_against_ur, candidate, vacancy_dict):
    if str(category) in ['3', '4', '5', '7', '8']:
    
        key_cat1 = str(candidate['cat1_cpo']) + str(post) + str(gender) + str(subcat)
        
        if (key_cat1 not in vacancy_dict.keys()) or (vacancy_dict[key_cat1]['initial_vacancy'] == 0):
            keyCat2= '9' + str(post) + str(gender) + str(subcat)
            
            if (keyCat2 not in vacancy_dict.keys()) or (vacancy_dict[keyCat2]['initial_vacancy'] == 0):
                return False
            else:
                if vacancy_dict[keyCat2]['allocated_hc'] != vacancy_dict[keyCat2]['initial_vacancy']:
                    vacancy_dict[keyCat2]['allocated_hc'] += 1
                allocated_against_ur = '1'
        else:
            if vacancy_dict[key_cat1]['allocated_hc'] != vacancy_dict[key_cat1]['initial_vacancy']:
                vacancy_dict[key_cat1]['allocated_hc'] += 1
            
            else :
                return False
                        
    candidates_df.loc[candidate.name, 'allocated_category'] = category
    candidates_df.loc[candidate.name, 'allocated_post'] = post
    candidates_df.loc[candidate.name, 'allocated_sub_category'] = subcat
    candidates_df.loc[candidate.name, 'allocated_against_ur'] = allocated_against_ur
    
    return True
```

```python
def allocate_candidates(candidates_df, vacancy_dict):
     
    filtered_candidates = candidates_df[(candidates_df['merit'].notnull())].sort_values(by='merit')
    
    for idx, candidate in filtered_candidates.iterrows():
        allocated = False
        roll = candidate['rollno']
        post_preference = candidate['post_pref']
        DOB = candidate['dob']
        gender = candidate['gender']
       
        
        for post in post_preference.split(','):

            if post == 'A':
                if candidate['cat2'] == '3':
                    if candidate['catsel_spclexs_dp'] != '':
                        subcat = '3*'
                        catsel_name = 'catsel_spclexs_dp'
                        catsel = candidate['catsel_spclexs_dp']
                    else:
                        subcat = '3'
                        catsel_name = 'catsel_dp'
                        catsel = candidate['catsel_dp']
                else:
                    subcat = '0'
                    catsel_name = 'catsel_dp'
                    catsel = candidate['catsel_dp']
            
                
                for category in catsel:

                    if catsel != candidate['catsel_spclexs_dp'] and category == '3':
                        category = candidate['cat1_dp']
                        subcat = '3'
                    elif catsel == candidate['catsel_spclexs_dp']:
                        category = candidate['cat1_dp']
                        subcat = '3*'
                    else:
                        subcat = '0'
                    
                    key =  str(category) + str(post) + str(gender) + str(subcat)
                    
                    allocated_against_ur = ''
                    
                    if key in vacancy_dict and vacancy_dict[key]['current'] > 0 and candidate['total'] >= vacancy_dict[key]['min_marks_old']:
                        post = vacancy_dict[key]['post_code']
                                    
                        allocated = update_allocation_dp(candidates_df, category, post, gender, subcat, allocated_against_ur, candidate, vacancy_dict)
                        if allocated:
                            vacancy_dict[key]['current'] -= 1
                            vacancy_dict[key]['allocated'] += 1
                            break
                    else:
                        if subcat == '3*' and vacancy_dict[key]['current'] == 0:
                            subcat = '3'
                            key = str(category) + str(post) + str(gender) + str(subcat)
                            if key in vacancy_dict and vacancy_dict[key]['current'] > 0 and candidate['total'] >= vacancy_dict[key]['min_marks_old']:
                                post = vacancy_dict[key]['post_code']
                                        
                                allocated = update_allocation_dp(candidates_df, category, post, gender, subcat, allocated_against_ur, candidate, vacancy_dict)
                                if allocated:
                                    vacancy_dict[key]['current'] -= 1
                                    vacancy_dict[key]['allocated'] += 1
                                    break
                if allocated:
                    break
            else:
                subcat='0'
                catsel_name = 'catsel_cpo'
                catsel = candidate['catsel_cpo']

                for category in catsel:
                    key =  str(category) + str(post) + str(gender) + str(subcat)
                    
                    allocated_against_ur = ''
                    
                    if key in vacancy_dict and vacancy_dict[key]['current'] > 0 and candidate['total'] >= vacancy_dict[key]['min_marks_old']:
                        post = vacancy_dict[key]['post_code']
                                    
                        allocated = update_allocation_cpo(candidates_df, category, post, gender, subcat, allocated_against_ur, candidate, vacancy_dict)
                        if allocated:
                            vacancy_dict[key]['current'] -= 1
                            vacancy_dict[key]['allocated'] += 1
                            break
                if allocated:
                    break
    

    for key, vacancy_info in vacancy_dict.items():
        if key.endswith('*') and (vacancy_info['current'] != 0):
            filter_condition = (candidates_df['allocated_category'].isnull()) & (candidates_df['cat2'] == '3') & (candidates_df['catsel_dp'].notnull()) & (candidates_df['post_pref'].str.contains('A')) & (candidates_df['cat1_dp'].astype(str) == str(vacancy_info['category_code']))
            filtered_candidates_spl_esm = candidates_df[filter_condition]
    
            for i in range(vacancy_info['current']):
                index = filtered_candidates_spl_esm.index[i]
                columns_to_update = ['allocated_category', 'allocated_post', 'allocated_sub_category', 'allocated_against_ur']
                values_to_update = [vacancy_info['category_code'], vacancy_info['post_code'], '3', '']
                
                candidates_df.loc[index, columns_to_update] = values_to_update
                vacancy_info['current'] -= 1
                vacancy_info['allocated'] += 1
                continue

                
    return vacancy_dict
```

```python
def adjust_vacancy(candidates_df, vacancy_df):
    vacancy_df['current'] = vacancy_df['initial_vacancy'] - vacancy_df['allocated_hc']
    vacancy_df['allocated'] = 0
    vacancy_df['left_vacancy'] = 0
        
    vacancy_df['allocated_hc_prev'] = vacancy_df['allocated_hc']
    vacancy_df['allocated_hc'] = 0

    return vacancy_df
```

```python
def find_lowest_marks(candidates_df, vacancy_df):
    candidates_df['key'] = (candidates_df['allocated_category'].astype(str) +
                            candidates_df['gender'].astype(str) +
                            candidates_df['allocated_post'].astype(str) +
                            candidates_df['allocated_sub_category'].astype(str))

    vacancy_df['key'] = (vacancy_df['category_code'].astype(str) +
                         vacancy_df['gender'].astype(str) +
                         vacancy_df['post_code'].astype(str) +
                         vacancy_df['sub_category_code'].astype(str))

    highest_merit_candidates = candidates_df.loc[candidates_df.groupby('key')['merit'].idxmax()]

    vacancy_df_merged = vacancy_df.merge(highest_merit_candidates[['key', 'total', 'paper2' , 'dob', 'merit']],
                                  on='key', how='left')

    vacancy_df_merged['min_marks'] = vacancy_df_merged['total']
    vacancy_df_merged['min_marks_paper2'] = vacancy_df_merged['paper2']
    #vacancy_df_merged['min_marks_paper1'] = vacancy_df_merged['paper1']
    vacancy_df_merged['min_marks_cand_dob'] = vacancy_df_merged['dob']
    vacancy_df_merged['min_marks_merit'] = vacancy_df_merged['merit']


    #vacancy_df.drop(columns=['total_marks', 'parta_gi', 'partb_ga', 'dob'], inplace=True)

    return vacancy_df_merged
```
