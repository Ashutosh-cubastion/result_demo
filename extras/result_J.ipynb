```python
def get_merit(candidate_df, medium):    
    if medium == 'C':
        post = ['CIVIL']
        merit = 'merit_c'
    elif medium == 'EM':
        post = ['ELECT', 'MECH']
        merit = 'merit_em'
    else:
        return "Enter Correct Medium i.e. Either 'C' or 'EM'"

    candidate_df[merit] = None
    
    candidates =  candidate_df[(candidate_df['total'].notnull()) & (candidate_df['rejection_provision'] == 'C') &
    (candidate_df['post_applied'].isin(post))].sort_values(by = ['total', 'paper2', 'dob', 'cand_name'], ascending=[False, False, True, True])
    
    candidates[merit] = range(1, len(candidates) + 1)

    rollno_to_merit = candidates.set_index('rollno')[merit].to_dict()
    candidate_df[merit] = candidate_df['rollno'].map(rollno_to_merit)
    
    return "{flag_name} updated successfully for {merit} candidates".format(
        flag_name=merit,
        merit=len(candidates))
```

### CUTOFF FLAG

#### CODE

```python
def get_cutoff(cutoff_df, candidate_df, medium):
    
    if medium == 'C':
        post = ['CIVIL']
        merit = 'merit_c'
        cutoff = 'cutoff_c'
    elif medium == 'EM':
        post = ['ELECT', 'MECH']
        merit = 'merit_em'
        cutoff = 'cutoff_em'
    else:
        return "Enter Correct Medium i.e. Either 'C' or 'EM'"

    
    candidate_df[cutoff] = None
    cutoff_dict = {}

    for index, row in cutoff_df[cutoff_df['flag'].astype(str) == cutoff].iterrows():
        category = int(row['category'])
        paper1 = float(row['paper1'])
        paper2 = float(row['paper2'])
        cutoff_dict[category] = [paper1, paper2]
    
    sorted_candidates = candidate_df[candidate_df[merit].notnull()].sort_values(by=merit)

    def calculate_cutoff(candidate):
        cutoff = ''

        if (candidate['paper1'] >= cutoff_dict[9][0]) & (candidate['paper2'] >= cutoff_dict[9][1]):
            cutoff += '9'

        for cat in [0, 1, 2, 6]:
            if (candidate['cat1'] == cat) & (cat in cutoff_dict):
                if (candidate['paper1'] >= cutoff_dict[cat][0]) & (candidate['paper2'] >= cutoff_dict[cat][1]):
                    cutoff += str(cat)

        if (candidate['cat2'] == 3) & (3 in cutoff_dict):
            if (candidate['paper1'] >= cutoff_dict[3][0]) & (candidate['paper2'] >= cutoff_dict[3][1]):
                cutoff += '3'

        for cat in [4, 5, 7, 8]:
            if (candidate['cat3'] == cat) & (cat in cutoff_dict):
                if (candidate['paper1'] >= cutoff_dict[cat][0]) & (candidate['paper2'] >= cutoff_dict[cat][1]):
                    cutoff += str(cat)

        return cutoff

    sorted_candidates[cutoff] = sorted_candidates.apply(calculate_cutoff, axis=1)
    candidate_df.update(sorted_candidates)

    return f"{cutoff} updated successfully"
```

```python
def get_dob_flag(candidate_df, arc_df, age_group):
    if age_group == '18_30':
        dob_to_date = pd.to_datetime('08/01/2006', format="%m/%d/%Y")
        dob_from_date = pd.to_datetime('08/01/1994', format="%m/%d/%Y")
        dob_flag = 'dob_'+ age_group
    elif age_group == '18_32':
        dob_to_date = pd.to_datetime('08/01/2006', format="%m/%d/%Y")
        dob_from_date = pd.to_datetime('08/01/1992', format="%m/%d/%Y")
    else:
        return "Enter Valid age group i.e. either 18_30 or 18_32"

    dob_flag = 'dob_'+ age_group

    arc_year_dict = dict(zip(arc_df['arc_code'].astype(str), arc_df['arc_year']))

    candidate_df[dob_flag] = ''

    def calculate_verification_flag(row):
        dob = row['dob']
        verification_flag = ''
        
        if dob >= dob_to_date:
            verification_flag = 'U'
        elif dob_from_date < dob < dob_to_date:
            verification_flag = '9'
        elif dob <= dob_from_date:
            agerelax_code = row['agerelax_code']
            
            if pd.isna(agerelax_code) or agerelax_code == '':
                agerelax_code = '99'
            
            if isinstance(agerelax_code, str) and len(agerelax_code.strip()) == 1:
                agerelax_code = agerelax_code.strip()
            
            age_rlx_year = arc_year_dict.get(agerelax_code, 0)
            
            if agerelax_code == '6':
                ex_service_years = row['exsm_yrs']
                ex_service_months = row['exsm_months']
                ex_service_days = row['exsm_days']
                age_rlx_year += int(ex_service_years) if ex_service_years else 0
                new_dob = dob + pd.DateOffset(years=age_rlx_year, months=ex_service_months, days=ex_service_days)
            
            else:
                extract_year = dob.year
                extract_month = dob.month
                extract_day = dob.day
                new_year = extract_year + int(age_rlx_year)
                
                if extract_month == 2 and extract_day == 29:
                    extract_day = 28
            
                new_dob = pd.to_datetime(f"{new_year}-{extract_month}-{extract_day}", format='%Y-%m-%d')
                
            if new_dob > dob_from_date:
                if len(agerelax_code) < 2:
                    verification_flag = '0' + agerelax_code
                else:
                    verification_flag = agerelax_code
            else:
                verification_flag = '99'

        return verification_flag

    sorted_candidates = candidate_df[(candidate_df['merit_c'].notnull()) | (candidate_df['merit_em'].notnull())]
    
    sorted_candidates[dob_flag] = sorted_candidates.apply(calculate_verification_flag, axis=1)

    candidate_df.update(sorted_candidates)
    
    return f"{dob_flag} updated successfully"
```

#### CODE

```python
def get_catsel_dob(candidate_df, arc_df, age_group):
    
    if age_group == '18_30':
        dob_from_date = pd.to_datetime('08/01/1994', format="%m/%d/%Y")
        
    elif age_group == '18_32':
        dob_from_date = pd.to_datetime('08/01/1992', format="%m/%d/%Y")
    else:
        return "Enter Valid age group i.e. either 18_30 or 18_32"

    dob_flag = 'dob_'+ age_group
    catsel_dob_flag = 'catsel_dob_'+ age_group

    candidate_df[catsel_dob_flag] = ''
    
    filtered_and_sorted_candidates = candidate_df[((candidate_df['merit_c'].notnull()) | (candidate_df['merit_em'].notnull())) 
    & (candidate_df[dob_flag].notnull()) 
    & (candidate_df[dob_flag] != '')]
    def calculate_catsel_dob(row, dob_flag_name):

        dob_flag = row[dob_flag_name]
       
        if dob_flag is not None and dob_flag != '' and dob_flag != '9':
            matching_rows = arc_df[arc_df['arc_code'] == dob_flag]
            if not matching_rows.empty:
                catsel = matching_rows.iloc[0]
                catsel_cat1 = catsel['arc_cat1'].split(',')
                catsel_cat2 = catsel['arc_cat2'].split(',')
                catsel_cat3 = catsel['arc_cat3'].split(',')
                catsel_gender = catsel['arc_gender'].split(',')
                condition = (str(row['cat1']) in catsel_cat1) & (str(int(row['gender'])) in catsel_gender)
                catsel_dob = ''
                if row['cat2'] == '3' and row['cat2'] in catsel_cat2:
                    catsel_dob = row['cat2']
                else:
                    if pd.isnull(catsel['arc_year_against_ur']):
                        catsel['arc_year_against_ur'] = 0
                    dob_datetime = pd.to_datetime(row['dob'])
                    extract_year = dob_datetime.year
                    extract_month = dob_datetime.month
                    extract_day = dob_datetime.day
                    if (extract_day == 29) and (extract_month == 2):
                        extract_day = 28
                    new_year = extract_year + int(catsel['arc_year_against_ur'])
                    new_dob = pd.to_datetime(f'{new_year}-{extract_month}-{extract_day}')
                    catsel_dob_decision = new_dob > dob_from_date
                    if (str(row['cat3']) in catsel_cat3 and str(row['cat3']) != ''):
                        if catsel_dob_decision:
                            catsel_dob = '9'
                        else:
                            catsel_dob = row['cat3']  
                    else:
                        catsel_dob = '9' if catsel_dob_decision else row['cat1']
                return catsel_dob if condition else ''
            else:
                return ''
        else:
            return dob_flag
    
    filtered_and_sorted_candidates[catsel_dob_flag] = filtered_and_sorted_candidates.apply(calculate_catsel_dob, args = (dob_flag, ), axis=1)
    candidate_df.update(filtered_and_sorted_candidates)
    
    print('Successfully updated ' + catsel_dob_flag)
```

### CATSEL FLAG

#### CODE

```python
def get_catsel(candidate_df, medium, age_group):
    
    if medium == 'C':
        if age_group == '18_30':
            catsel_dob_flag = 'catsel_dob_'+age_group
            cutoff_flag = 'cutoff_c'
            merit = 'merit_c'
        elif age_group == '18_32':
            catsel_dob_flag = 'catsel_dob_'+age_group
            cutoff_flag = 'cutoff_c'
            merit = 'merit_c'
        else:
            return "Incorrect Age Group"
    elif medium == 'EM':
        if age_group == '18_30':
            catsel_dob_flag = 'catsel_dob_'+age_group
            cutoff_flag = 'cutoff_em'
            merit = 'merit_em'
        elif age_group == '18_32':
            catsel_dob_flag = 'catsel_dob_'+age_group
            cutoff_flag = 'cutoff_em'
            merit = 'merit_em'
        else:
            return "Incorrect Age Group"
    else:
        return "Incorrect medium"


    catsel_flag = 'catsel_'+medium.lower()+'_'+age_group
    candidate_df[catsel_flag] = ''
    
    candidates = candidate_df[candidate_df[merit].notnull()].sort_values(by='rollno')
    
    def update_cutoff_flag(candidate_record):
        if candidate_record['cat2'] == '3' and candidate_record['exs_reservation'] == 'No':
            candidate_record[cutoff_flag] = candidate_record[cutoff_flag].replace('3', '').strip()

        if candidate_record[catsel_dob_flag] in ['1', '2', '6', '4', '5', '7', '8']:
            return candidate_record[cutoff_flag].replace('9', '').strip()
        elif pd.isna(candidate_record[catsel_dob_flag]) or candidate_record[catsel_dob_flag] == '':
            return ''
        else:
            return candidate_record[cutoff_flag]
    
    candidate_df[catsel_flag] = candidates.apply(update_cutoff_flag, axis=1)

    return 'Successfully updated '+ catsel_flag
```

```python
def update_allocation_c(candidates_df, category, post, subject, allocated_against_ur, candidate, vacancy_dict):
    if str(category) in ['3', '4', '5', '7', '8']:
    
        key_cat1 = str(candidate['cat1']) + str(subject) + str(post)
        
        if (key_cat1 not in vacancy_dict.keys()) or (vacancy_dict[key_cat1]['initial_vacancy'] == 0):
            keyCat2='9' + str(subject) + str(post)
            
            if (keyCat2 not in vacancy_dict.keys()) or (vacancy_dict[keyCat2]['initial_vacancy'] == 0):
                return False
            else:
                if vacancy_dict[keyCat2]['allocated_hc'] != vacancy_dict[keyCat2]['initial_vacancy']:
                    vacancy_dict[keyCat2]['allocated_hc'] += 1
                allocated_against_ur = '1'
        else:
            if vacancy_dict[key_cat1]['allocated_hc'] != vacancy_dict[key_cat1]['initial_vacancy']:
                vacancy_dict[key_cat1]['allocated_hc'] += 1
            
            else :
                return False
                        
    candidates_df.loc[candidate.name, 'allocated_subject'] = subject
    candidates_df.loc[candidate.name, 'allocated_category'] = category
    candidates_df.loc[candidate.name, 'allocated_post'] = post
    candidates_df.loc[candidate.name, 'allocated_against_ur'] = allocated_against_ur
    
    return True
```

```python
def update_allocation_em(candidates_df, category, post, subject, allocated_against_ur, candidate, vacancy_dict):
    if str(category) in ['3', '4', '5', '7', '8']:
        
        key_cat1 = str(candidate['cat1']) + subject + str(post)
        
        if (key_cat1 not in vacancy_dict.keys()) or (vacancy_dict[key_cat1]['initial_vacancy'] == 0):
            keyCat2='9' + subject + str(post)
            
            if (keyCat2 not in vacancy_dict.keys()) or (vacancy_dict[keyCat2]['initial_vacancy'] == 0):
                return False
            else:
                if vacancy_dict[keyCat2]['allocated_hc'] != vacancy_dict[keyCat2]['initial_vacancy']:
                    vacancy_dict[keyCat2]['allocated_hc'] += 1
                allocated_against_ur = '1'
        else:
            if vacancy_dict[key_cat1]['allocated_hc'] != vacancy_dict[key_cat1]['initial_vacancy']:
                vacancy_dict[key_cat1]['allocated_hc'] += 1
            
            else :
                return False
                        
    candidates_df.loc[candidate.name, 'allocated_subject'] = subject
    candidates_df.loc[candidate.name, 'allocated_category'] = category
    candidates_df.loc[candidate.name, 'allocated_post'] = post
    candidates_df.loc[candidate.name, 'allocated_against_ur'] = allocated_against_ur
    
    return True
```

```python
def allocate_candidates(candidates_df, vacancy_dict, medium):

    if medium == 'C':
        
        merit = 'merit_c'    
        filtered_candidates = candidates_df[(candidates_df['merit_c'].notnull()) & 
        (candidates_df['allocated_category'].isnull())].sort_values(by='merit_c')
        subject_option = 'C'
    
    elif medium == 'EM':
        
        merit = 'merit_em'
        filtered_candidates = candidates_df[(candidates_df['merit_em'].notnull()) & 
        (candidates_df['allocated_category'].isnull())].sort_values(by='merit_em')
        
    else:
        return "Enter Correct Medium i.e. Either 'C' or 'EM'"

    post_flag_dict = {'BFGIKM':['catsel_dob_18_30', 'catsel_c_18_30'],
                      'AEHJLNO':['catsel_dob_18_30', 'catsel_em_18_30'],
                      'C':['catsel_dob_18_32', 'catsel_em_18_32'],
                      
                      'D':['catsel_dob_18_32', 'catsel_c_18_32']}
    
    
    for idx, candidate in filtered_candidates.iterrows():
        allocated = False
        roll = candidate['rollno']
        post_preference = candidate['post_pref']
        print(post_preference)
        DOB = candidate['dob']
        
        for post in post_preference.split(','):

            if post == 'X':
                continue
        
            catsel_dob_name = [values for key, values in post_flag_dict.items() if post in key][0][0]
            catsel_name = [values for key, values in post_flag_dict.items() if post in key][0][1]

            catsel_dob = str(candidate[catsel_dob_name])
            catsel = str(candidate[catsel_name])
            
            if medium == "C" and catsel != '':

                if post == 'B' and (candidate['gender'] == 1):
                    continue
                    
                subject_option = "C"
                for category in catsel:
                    key_c = str(category) + subject_option + str(post)
                    allocated_against_ur = ''
                    if key_c in vacancy_dict and vacancy_dict[key_c]['current'] > 0:
                                    
                        allocated = update_allocation_c(candidates_df, category, post, subject_option, allocated_against_ur, candidate, vacancy_dict)
                        if allocated:
                            vacancy_dict[key_c]['current'] -= 1
                            vacancy_dict[key_c]['allocated'] += 1
                            break
                if allocated:
                    break

            if medium == "EM" and catsel != '':

                if post == 'A' and (candidate['gender'] == 1):
                    continue
                
                if post in ('A', 'J', 'C'):
                    subject_option  = "EM"
                elif post in ('H', 'L', 'N'):
                    subject_option  = "E"
                else:
                    subject_option = 'M'
                        
                for category in catsel:
                    key_em = str(category) + subject_option + str(post)
                    allocated_against_ur = ''
                    
                    if key_em in vacancy_dict and vacancy_dict[key_em]['current'] > 0:            
                        allocated = update_allocation_em(candidates_df, category, post, subject_option, allocated_against_ur, candidate, vacancy_dict)
                        
                        if allocated:
                            vacancy_dict[key_em]['current'] -= 1
                            vacancy_dict[key_em]['allocated'] += 1
                            
                            break
                if allocated:
                    break

    return vacancy_dict
```

```python
def update_vacancy_table(vacancy_table_df, category_code, subject_option, post_code, vacancy):
    condition = (
        (vacancy_table_df['category_code'].astype(str) == category_code) &
        (vacancy_table_df['subject_option'].astype(str) == subject_option) &
        (vacancy_table_df['post_code'].astype(str) == post_code)
    )
    if condition.any():
        current = vacancy_table_df.loc[condition, 'current']
        vacancy_table_df.loc[condition, 'current'] -= vacancy


def adjust_vacancy(candidates_df, vacancy_df):
    vacancy_df['current'] = vacancy_df['initial_vacancy']
    vacancy_df['allocated'] = 0
    vacancy_df['left_vacancy'] = 0

    grouped_candidates = candidates_df[candidates_df['allocated_category'].isin(['3', '4', '5', '8'])].groupby(['cat1', 'allocated_subject', 'allocated_post', 'allocated_against_ur'])

    for (category_code, subject_option, post_code, allocated_against_ur), group in grouped_candidates:
        category_code = '9' if allocated_against_ur == '1' else category_code
        vacancy = len(group)
        
        update_vacancy_table(vacancy_df, category_code, subject_option, post_code, vacancy)
        
    vacancy_df['allocated_hc_prev'] = vacancy_df['allocated_hc']
    vacancy_df['allocated_hc'] = 0

    return vacancy_df
```

```python
def find_lowest_marks(candidates_df, vacancy_df):
    vc = []
    vacancy_df['min_total'] = 0
    for _, row in vacancy_df.iterrows():
        lmv = {
            "category_code": str(row["category_code"]),
            "subject_option": str(row["subject_option"]),
            "post_code": str(row["post_code"]),
        }

        vc.append(lmv)

    print("Vector size--->", len(vc))

    for lmv1 in vc:
        highest_merit_candidate = get_highest_merit_candidate(candidates_df, lmv1["category_code"], lmv1["subject_option"], lmv1["post_code"])
        update_vacancy_table(vacancy_df, lmv1["category_code"], lmv1["subject_option"], lmv1["post_code"], highest_merit_candidate)

def get_highest_merit_candidate(candidates_df, category_code, subject_code, post_code):
    highest_merit_candidate = None

    try:
        mask = (
            (candidates_df["allocated_category"] == str(category_code))
            & (candidates_df["allocated_subject"] == str(subject_code))
            & (candidates_df["allocated_post"] == str(post_code))
        )

        filtered_cand = candidates_df[mask]

        # Assuming 'merit_c' or 'merit_em' is the column representing the merit of the candidate
        if str(subject_code) == 'C':
            highest_merit_candidate = filtered_cand.loc[filtered_cand['merit_c'].idxmax()]
        else:
            highest_merit_candidate = filtered_cand.loc[filtered_cand['merit_em'].idxmax()]

    except Exception as ex:
        print(ex)

    return highest_merit_candidate

def update_vacancy_table(vacancy_df, category_code, subject_code, post_code, highest_merit_candidate):
    try:
        key = str(category_code) + str(subject_code) + str(post_code)
        mask = (vacancy_df['key'] == key)
        vacancy = vacancy_df.loc[mask]

        vacancy["min_total"] = highest_merit_candidate['total']
        vacancy["min_paper2"] = highest_merit_candidate['paper2']
        vacancy["min_cand_dob"] = highest_merit_candidate['dob']
        vacancy["min_paper1"] = highest_merit_candidate['paper1']

        # Use .item() to get a scalar value from the Series
        if vacancy['subject_option'].item() == 'C':
            vacancy['max_merit_civil'] = highest_merit_candidate['merit_c']
        else:
            vacancy['max_merit_em'] = highest_merit_candidate['merit_em']

        vacancy_df.loc[mask, 'min_total'] = vacancy["min_total"]
        vacancy_df.loc[mask, 'min_paper2'] = vacancy["min_paper2"]
        vacancy_df.loc[mask, 'min_cand_dob'] = vacancy["min_cand_dob"]
        vacancy_df.loc[mask, 'min_paper1'] = vacancy["min_paper1"]
        vacancy_df.loc[mask, 'max_merit_civil'] = vacancy["max_merit_civil"]
        vacancy_df.loc[mask, 'max_merit_em'] = vacancy["max_merit_em"]

        # Print statements for verification
        
    except Exception as ex:
        print(ex)
```
